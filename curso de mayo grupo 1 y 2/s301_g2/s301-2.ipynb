## Paso 1 - Cargar los datos CSV de la financiera

```python
import pandas

financiera = pandas.read_csv("financiera.csv")

financiera
```

```python
financiera.info()
```

```python
financiera[["folio", "numero", "pagoFechaEsperada", "pagoFechaRecibido", "pagoMontoEsperado", "pagoMontoRecibido"]].info()
```

```python
financiera["montoFijadoPagado"] = financiera["pagoMontoEsperado"]
financiera["montoEsperadoPagado"] = financiera["montoFijadoPagado"] * financiera["pagado"]

financiera
```

```python
avanceVentas = financiera.groupby("folio").agg({
    "pagoMontoEsperado": "sum",
    "pagoMontoRecibido": "sum",
    "montoFijadoPagado": "sum",
    "montoEsperadoPagado": "sum",
    "folio": "count",
    "precio": "first",
    "anticipo": "first",
    "semanal": "first",
    "total": "first",
})

avanceVentas
```

```python
avanceVentas.to_csv("financiera_fact_avance_ventas.csv")
```
